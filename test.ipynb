Two sum

```python
nums=[2,7,11,15]
target=9
l=0
r=len(nums)-1
for i in range(len(nums)):
    if nums[l]+nums[r]==target:
        print([l,r])
        break
    elif nums[l]+nums[r]<target:
        l+=1
    else:
        r-=1
```

Reverse number

```python
x=121
if str(x)==str(x)[::-1]:
    print(True)
else:
    print(False)
```

Duplicate element 

```python
nums = [1, 2, 3,4,77,34,90,23,77] 
res=[]
for i in nums:
    if i in res:
        print(True)
    else:
        res.append(i)
```

Prime number

```python
n=29
if n<=1:
    print(False)
else:
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            print(False)
            break
    else:
        print(True)
```

Reverse String

```python
s = "A man, a plan, a canal: Panama"

new = ""

for c in s:
    if c.isalpha():
        new += c.lower()

print(new)
if new==new[::-1]:
    print(True)
else:
    print(False)
```

Number belongs to fibonacci series


```python
n=int(input())
res=[]
a,b=0,1
for i in range(n):
    res.append(a)
    a,b=b,a+b
if n in res:
    print(True)
else:
    print(False)
    
```

Missing number

```python
nums=[1,8,27,64,216]
nums2=[]
for i in range(1,7):
    nums2.append(i**3)
ans=sum(nums2)-sum(nums)
print(ans)
```

List function

```python
nums=[2,5,8,3,10,7]
res=[]
for i in nums:
    if i%2==0:
        res.append(i//2)
    else:
        res.append(i*2)
print(res)
```
